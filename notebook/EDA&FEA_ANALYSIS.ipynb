```python
df.columns = df.columns.str.strip()
df.columns
```

```python
## numerical features
num_features = [fea for fea in df.columns if df[fea].dtype != 'object']

## catagorical features
cat_features = [fea for fea in df.columns if df[fea].dtype == 'object']
```

```python
print(f'Total Numerical features are {len(num_features)} : {num_features}')
print(f'Total Catagorical features are {len(cat_features)} : {cat_features}')
```

```python
##value counts

for i in cat_features:
    print(df[i].value_counts(),'\n')
```
